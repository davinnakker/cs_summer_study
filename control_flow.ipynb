```python
def process_command(command: list[str]) -> str:
    match command:
        case ["move", direction, distance] if int(distance) > 0:
            return f"Moving {direction} by {distance}, units"
        case ["attack", target]:
            return f"Attacking {target}"
        case ["heal"]:
            return "Healing..."
        case _:
            return "Unknown Command"

process_command(["move", "North", "10"])
```

```python
def handle_event(event: dict) -> str:
    match event:
        case {"type": "user_created", "user": {"id": user_id, "name": name}}:
            return f"Created user {name} with ID {user_id}"
        case {"type": "user_updated", "user": {"id": _, "name": name}, "fields": fields}:
            return f"Updated user {name} (fields: {', '.join(fields)})"
        case {"type": "user_deleted", "user_id": user_id}:
            return f"Deleted user {user_id}"
        case {"type": "login_failed", "user_id": user_id, "reason": reason}:
            return f"Login failed for user {user_id}: {reason}"
        case _:
            return "Unknown event"

arg = {
    "type": "user_updated",
    "user": {
        "id": 42,
        "name": "Alice"
    },
    "fields": ["name", "email"]
}

handle_event(arg)
```
